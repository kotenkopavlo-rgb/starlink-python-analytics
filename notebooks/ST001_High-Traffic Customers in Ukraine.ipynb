## Task 1 — High-Traffic Customers in Ukraine

### Business Question

Which Ukrainian customers have downloaded more than 2000 GB?

### Analysis

Filter Ukrainian customers with more than 2000 GB downloaded and sort them by downloaded traffic in descending order.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
customers.loc[
    (customers["gb_downloaded"] > 2000) & (customers["country"] == "Ukraine"),
    ["unique_id","first_name","city","gb_downloaded"]
].sort_values(
    "gb_downloaded",
    ascending=False
)
```

### Result

277 customers meet the criteria.

### Business Insight

The analysis shows that 277 Ukrainian customers downloaded more than 2000 GB.
These customers represent 2.77% of the total customer base. This high-usage customer segment may be a good target for a special offer or a tailored subscription plan.
