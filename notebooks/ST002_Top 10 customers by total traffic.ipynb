## Task 2 — Top 10 customers by total traffic

### Business Question

Who are the top 10 customers by total traffic?

### Analysis

Calculate total traffic for each customer, then sort the customers by total traffic in descending order and select the top 10.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
customers["total_gb"] = customers["gb_downloaded"] + customers["gb_uploaded"]
customers[["unique_id","first_name","country","gb_downloaded","gb_uploaded","total_gb"]].sort_values(
    "total_gb",
    ascending=False
).head(10)
```

### Result

The top 10 customers have total traffic ranging from 11,570 GB to 14,473 GB.

### Business Insight

The top 10 customers generate very high traffic, which may indicate that they are among our most valuable customers. We could offer them a special deal or discount to encourage retention.
