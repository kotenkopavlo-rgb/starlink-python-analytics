## Task 3 — Traffic by Country

### Business Question

Which countries have the highest average download traffic?

### Analysis

Group customers by country, calculate the average download traffic for each country, sort the results in descending order, and select the top 5 countries.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
result = customers.groupby("country")["gb_downloaded"].mean().sort_values(
    ascending=False
).head(5).reset_index()
result = result.rename(columns={"gb_downloaded": "average_download_gb"})
result
```

### Result

The United States has the highest average download traffic at 1,573 GB per customer, while Spain has the lowest among the top 5 countries at 1,526 GB per customer.

### Business Insight

There is no significant difference in average download traffic among the top 5 countries.
