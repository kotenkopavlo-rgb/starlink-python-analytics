## Task 4  — Country Traffic & Customer Analysis

### Business Question

Which countries have the highest number of customers and the highest total download traffic?

### Analysis

Group customers by country, calculate the number of customers, average download traffic, and total download traffic for each country. Sort the results by total download traffic in descending order and select the top 5 countries.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
customers.groupby("country")["unique_id"].count()
customers.groupby("country")["gb_downloaded"].mean()
customers.groupby("country").agg({
    "unique_id": "count",
    "gb_downloaded": ["mean", "sum"]
}).sort_values(
    ("gb_downloaded","sum"),
    ascending=False
).head(5)
```

### Result

Ukraine has the highest total download traffic among all countries at 1,612,121.82 GB, followed by Germany (1,575,962.05 GB) and Australia (1,562,854.14 GB). The top 5 countries have between 977 and 1,032 customers each.

### Business Insight

The top 5 countries have similar customer counts and average download traffic, so differences in total traffic are mainly driven by the number of customers and their usage levels.
