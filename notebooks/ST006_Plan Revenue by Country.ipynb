## Task 6 — Plan Revenue by Country

### Business Question

Which subscription plan generates the highest monthly revenue in each country?

### Analysis

Group customers by country and subscription plan, calculate the total monthly revenue for each combination, sort the results by total revenue in descending order, and identify the highest-revenue plan for each country.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
revenue_by_plan_and_country = starlink.groupby(["country","plan_name"]).agg(
    total_revenue = ("monthly_fee","sum")
).sort_values(
    "total_revenue",
    ascending = False
).groupby("country").head(1)
print(revenue_by_plan_and_country)
```

### Result

The Residential plan generates the highest total monthly revenue in all analyzed countries. Australia has the highest Residential revenue at 77,280, followed by Germany at 75,600 and Canada at 73,920.

### Business Insight

The Residential plan generates the highest total monthly revenue in every country. This indicates that Residential is the main revenue-generating subscription plan across all analyzed markets. However, the analysis does not show whether this is caused by a higher number of Residential customers, a higher subscription fee, or both.
