## Task 11 — Revenue by Country

### Business Question

Which countries generate the highest average monthly revenue per customer?

### Analysis

Group customers by country, calculate total monthly revenue and the number of customers for each country, then calculate average monthly revenue per customer and sort the countries in descending order by this metric.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
total_revenue_by_country = starlink.groupby("country").agg(
    total_revenue = ("monthly_fee","sum"),
    users_amount = ("unique_id","count")
)
total_revenue_by_country["average_revenue_per_customer"] = total_revenue_by_country["total_revenue"] / total_revenue_by_country["users_amount"]
total_revenue_by_country_sorted = total_revenue_by_country.sort_values(
    "average_revenue_per_customer",
    ascending = False
)
print(total_revenue_by_country_sorted)
```

### Result

Japan has the highest average monthly revenue per customer at 120.98, followed by Canada at 119.99 and Spain at 119.82. Germany has the lowest average revenue per customer at 115.43.

### Business Insight

Average revenue per customer varies only moderately across countries, despite differences in total revenue. Japan generates the highest revenue per customer, while Australia generates the highest total revenue due to its larger customer base. This indicates that customer volume has a stronger impact on total revenue than small differences in average revenue per customer.
