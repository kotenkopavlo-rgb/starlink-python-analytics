## Task 12 — Customer Activity by Country

### Business Question

Which countries have the highest average customer activity?

### Analysis

Group customers by country, calculate the number of customers, average download traffic, and average loyalty points for each country, then sort the results by average download traffic in descending order.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
customer_activity = starlink.groupby("country").agg(
    users_amount = ("unique_id","count"),
    gb_downloaded_average = ("gb_downloaded","mean"),
    loyalty_points_average = ("loyalty_points","mean")
).sort_values(
    "gb_downloaded_average",
    ascending = False
)
print(customer_activity)
```

### Result

The United Kingdom has the highest average download traffic at 1,573.76 GB per customer, followed by Ukraine at 1,562.13 GB and Germany at 1,534.53 GB. Loyalty points data is unavailable for most countries due to missing values.

### Business Insight

Average download traffic is relatively high across all analyzed countries, with the United Kingdom showing the highest value. The limited availability of loyalty points data prevents a meaningful comparison of customer loyalty across countries and indicates that this metric requires more complete data before further analysis.
