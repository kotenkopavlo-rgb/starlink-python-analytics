## Task 13 — Upload Traffic by Country

### Business Question

Which countries have the highest average upload traffic per customer?

### Analysis

Group customers by country, calculate the number of customers and total upload traffic for each country, then calculate the average upload traffic per customer and sort the results in descending order.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
upload_traffic_by_country = starlink.groupby("country").agg(
    users_amount = ("unique_id","count"),
    total_gb_uploaded = ("gb_uploaded","sum")
)
upload_traffic_by_country["average_gb_uploaded_per_user"] = upload_traffic_by_country["total_gb_uploaded"] / upload_traffic_by_country["users_amount"]
upload_traffic_by_country_sorted = upload_traffic_by_country.sort_values(
    "average_gb_uploaded_per_user",
    ascending = False
)
print(upload_traffic_by_country_sorted)
```

### Result

Ukraine has the highest average upload traffic per customer at 268.89 GB, followed by the United Kingdom at 268.06 GB and Spain at 263.54 GB. Japan has the lowest average upload traffic at 251.95 GB per customer.

### Business Insight

Average upload traffic varies moderately across countries, with Ukraine showing the highest value. The relatively small difference between the countries suggests that upload activity is fairly consistent across the analyzed markets.
