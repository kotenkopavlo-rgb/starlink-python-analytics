## Task 16 — Average Traffic by Subscription Plan

### Business Question

Which subscription plans have the highest average download traffic?

### Analysis

Group customers by subscription plan, calculate the number of customers and average download traffic for each plan, and sort the results in descending order by average download traffic.

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
customer_count_by_country = starlink.groupby("country").agg(
    users_amount = ("unique_id","count"),
).sort_values(
    "users_amount",
    ascending = False
)
print(customer_count_by_country)
```

### Result

The Business plan has the highest average download traffic at 1,546.63 GB per customer, followed by Roam at 1,531.22 GB and Residential at 1,507.88 GB. Mini has the lowest average download traffic at 1,503.51 GB per customer.

### Business Insight

Business customers have the highest average download traffic among all subscription plans, while Mini customers have the lowest. The difference between the plans is relatively small, indicating that average download activity is fairly similar across subscription segments.
