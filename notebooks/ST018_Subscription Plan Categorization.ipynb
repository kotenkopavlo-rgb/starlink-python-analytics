# Task 18 — Subscription Plan Categorization

## Business Question

How can we group subscription plans into broader business categories?

## Analysis

Map each subscription plan to a broader business category using a predefined mapping.


```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
plan_name_mapping = {
    "Residential" : "Home",
    "Mini" : "Budget",
    "Roam" : "Mobile",
    "Business" : "Commercial"
}
starlink["plan_category"] = starlink["plan_name"].map(plan_name_mapping)
print(starlink[["plan_name", "plan_category"]].head(10))
```

### Result

The subscription plans were successfully categorized into four broader groups:
- Residential → Home
- Roam → Mobile
- Business → Commercial
- Mini → Budget

### Business Insight

Grouping subscription plans into broader categories makes it easier to analyze customer segments and compare different types of services.
