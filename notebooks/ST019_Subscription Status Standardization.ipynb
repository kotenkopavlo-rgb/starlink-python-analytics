# Task 19 — Subscription Status Standardization

## Business Question

How can we standardize subscription statuses for reporting?

## Analysis

Map the existing subscription statuses to standardized, human-readable categories using `replace()` while preserving the original status column.

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
status_mapping = {
    "paused" : "Paused",
    "active" : "Active",
    "cancelled" : "Cancelled"
}
starlink["status_category"] = starlink["status"].replace(status_mapping)
print(starlink[["status","status_category"]].head(10))
starlink["status_category"].value_counts()
```

### Result

The subscription statuses were successfully standardized into Active, Paused, and Cancelled categories.

### Business Insight

Standardized status categories make subscription data easier to use in reporting and analysis while preserving the original status values.
