# Task 21 — Customer Download Usage Segmentation

## Business Question

How are customers distributed across different download usage categories?

## Analysis

Customers were segmented based on gb_downloaded:

0–1000 GB → Low Usage
1000–2000 GB → Normal Usage
2000+ GB → High Usage

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
starlink["download_category"] = pd.cut(
    starlink["gb_downloaded"],
    bins = [0, 1000, 2000, float("inf")],
    labels = ["Low usage", "Normal usage", "High usage"]
)
starlink["download_category"].value_counts()
```

### Result

Low Usage — 4,779
High Usage — 2,628
Normal Usage — 2,594

### Business Insight

Most customers belong to the Low Usage segment, while High and Normal Usage segments are relatively similar in size.
