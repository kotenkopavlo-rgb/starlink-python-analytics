# Task 25 — Customer Usage vs Country Average

## Business Question

Are customers using more or less data than the average customer in their country?

## Analysis

Calculated the average gb_downloaded for each country using groupby() and transform("mean"). Then calculated each customer's deviation from the country average and classified customers into three segments:

Above Average — usage is higher than the country average
Below Average — usage is lower than the country average
Average — usage is equal to the country average

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
starlink["country_avg_download"] = (
    starlink.groupby("country")["gb_downloaded"]
    .transform("mean")
)
starlink["download_vs_country_avg"] = (
    starlink["gb_downloaded"] - starlink["country_avg_download"]
)
conditions = [
    starlink["download_vs_country_avg"] < 0,
    starlink["download_vs_country_avg"] == 0,
    starlink["download_vs_country_avg"] > 0,
]
choices = ["Below Average", "Average", "Above Average"]
starlink["usage_vs_country"] = np.select(conditions, choices, default = "Unknown")
print(
    starlink[
        ["country", "gb_downloaded", "country_avg_download",
         "download_vs_country_avg", "usage_vs_country"]
    ].head(10)
)
```

### Business Insight

Customer data usage varies significantly within countries. Comparing individual customers with their country average helps identify customers with unusually high or low data consumption and provides additional context for analyzing customer behavior.
