# Task 26 — Customer Revenue Segment

## Business Question

How can customers be segmented based on their monthly subscription fee?

## Analysis

Segmented customers into three groups based on their monthly_fee:

High Value — monthly fee ≥ 200
Medium Value — monthly fee ≥ 100 and < 200
Low Value — monthly fee < 100

Customers without a subscription were assigned to Unknown.

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
starlink["country_avg_download"] = (
    starlink.groupby("country")["gb_downloaded"]
    .transform("mean")
)
starlink["download_vs_country_avg"] = (
    starlink["gb_downloaded"] - starlink["country_avg_download"]
)
conditions = [
    starlink["download_vs_country_avg"] < 0,
    starlink["download_vs_country_avg"] == 0,
    starlink["download_vs_country_avg"] > 0,
]
choices = ["Below Average", "Average", "Above Average"]
starlink["usage_vs_country"] = np.select(conditions, choices, default = "Unknown")
print(
    starlink[
        ["country", "gb_downloaded", "country_avg_download",
         "download_vs_country_avg", "usage_vs_country"]
    ].head(10)
)
```

### Business Insight

Customer segmentation based on monthly subscription fees helps identify customers by their potential revenue contribution. This segmentation can be used for further analysis of customer behavior, subscription plans, and revenue distribution.
