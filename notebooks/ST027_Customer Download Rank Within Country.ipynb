# Task 27 — Customer Download Rank Within Country

## Business Question

What is each customer's rank based on data usage within their country?

## Analysis

Analysis

Ranked customers within each country by gb_downloaded in descending order using groupby() and rank().

Customers with higher data consumption receive a higher position (rank 1 represents the highest usage). When multiple customers have the same gb_downloaded value, they receive the average of the corresponding ranks.

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
starlink["country_download_rank"] = (
    starlink.groupby("country")["gb_downloaded"]
    .rank(ascending = False)
)
print(
    starlink[
        ["country", "gb_downloaded", "country_download_rank"]
    ]
    .sort_values(["country", "country_download_rank"])
    .head(20)
)
```

### Business Insight

Country-level ranking helps identify customers with the highest and lowest data consumption relative to other customers in the same country. This can be used to analyze customer usage patterns and identify high-usage customers within each market.
