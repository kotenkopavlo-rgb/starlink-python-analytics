# Task 28 — High Usage Customers by Country

## Business Question

How many customers in each country use more than 3,000 GB of downloaded data?

## Analysis

Filtered customers with gb_downloaded > 3000, then grouped the results by country and counted the number of customers in each country. The results were sorted in descending order by the number of high-usage customers.

## Code

```python
import pandas as pd
import numpy as np
customers = pd.read_csv("starlink_customers.csv")
subscriptions = pd.read_csv("subscriptions.csv")
starlink = customers.merge(
    subscriptions,
    left_on = "unique_id",
    right_on = "customer_id",
    how = "left"
)
starlink_high_usage = (
    starlink[starlink["gb_downloaded"] > 3000]
    .groupby("country")
    .agg(
        high_usage_customers = ("unique_id","nunique")
    )
    .sort_values("high_usage_customers", ascending = False)
)
print(starlink_high_usage)
```

### Business Insight

The analysis identifies countries with the highest number of high-usage customers. This can help highlight markets with greater network demand and support further analysis of customer usage patterns and capacity requirements.
